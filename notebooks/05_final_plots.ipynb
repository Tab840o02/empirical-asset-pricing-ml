# 05 Final Plots

Phase 6 notebook for publication-quality figure export.

Running this notebook writes 300 dpi figures to report/figures.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
from src.config import PROCESSED_DIR

plt.rcParams.update({
    'figure.dpi': 120,
    'savefig.dpi': 300,
    'font.size': 11,
    'axes.grid': True,
    'grid.alpha': 0.25,
    'axes.spines.top': False,
    'axes.spines.right': False,
})

processed = Path(PROCESSED_DIR)
fig_dir = Path('report/figures')
fig_dir.mkdir(parents=True, exist_ok=True)

r2 = pd.read_csv(processed / 'eval_oos_r2_latest.csv', index_col=0).sort_values('oos_r2', ascending=False)
ext = pd.read_csv(processed / 'eval_ext_oos_r2.csv').pivot(index='model', columns='period', values='oos_r2')
pars = pd.read_csv(processed / 'eval_parsimony_oos_r2.csv', index_col=0)
deg = pd.read_csv(processed / 'eval_parsimony_degenerate_months_summary.csv')
print('Export path:', fig_dir.resolve())
```

```python
fig, ax = plt.subplots(figsize=(8.5, 4.2))
r2['oos_r2'].plot(kind='bar', ax=ax, color='#1f77b4')
ax.set_title('Phase 4 Pooled OOS R2 by Model')
ax.set_ylabel('OOS R2')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
fig.savefig(fig_dir / 'fig_phase4_oosr2.png', bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5.8))
im = ax.imshow(ext.values.astype(float), aspect='auto', interpolation='nearest')
ax.set_title('Phase 5a OOS R2 by Sub-Period')
ax.set_xticks(range(len(ext.columns)))
ax.set_xticklabels(ext.columns, rotation=45, ha='right')
ax.set_yticks(range(len(ext.index)))
ax.set_yticklabels(ext.index)
plt.colorbar(im, ax=ax, label='OOS R2')
plt.tight_layout()
fig.savefig(fig_dir / 'fig_phase5a_oosr2_heatmap.png', bbox_inches='tight')
plt.show()
```

```python
pars_sorted = pars.sort_values('delta_r2')
fig, ax = plt.subplots(figsize=(8.8, 4.2))
ax.bar(pars_sorted.index, pars_sorted['delta_r2'], color='#9467bd')
ax.set_title('Phase 5c Parsimony delta R2 (phase5c - phase4)')
ax.set_ylabel('delta R2')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
fig.savefig(fig_dir / 'fig_phase5c_delta_r2.png', bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(deg['model'], deg['zero_var_months'], color='#d62728')
ax.set_title('Phase 5c Degenerate Prediction Months by Model')
ax.set_ylabel('Zero-variance months')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
fig.savefig(fig_dir / 'fig_phase5c_degenerate_months.png', bbox_inches='tight')
plt.show()

print('Saved files:')
for f in sorted(fig_dir.glob('fig_phase*.png')):
    print('-', f.as_posix())
```
