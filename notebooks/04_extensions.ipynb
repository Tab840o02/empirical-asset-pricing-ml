# 04 Extensions

Phase 6 notebook for Extension 1, 2, and 3 summary visualizations (post-2020, transaction costs, parsimony).

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from src.config import PROCESSED_DIR

processed = Path(PROCESSED_DIR)
ext_r2 = pd.read_csv(processed / 'eval_ext_oos_r2.csv')
tc4 = pd.read_csv(processed / 'eval_tc_summary.csv')
tc_ext = pd.read_csv(processed / 'eval_tc_ext_summary.csv')
pars = pd.read_csv(processed / 'eval_parsimony_oos_r2.csv', index_col=0)
deg = pd.read_csv(processed / 'eval_parsimony_degenerate_months_summary.csv')
print('Extension files loaded.')
```

```python
heat = ext_r2.pivot(index='model', columns='period', values='oos_r2')
fig, ax = plt.subplots(figsize=(10, 6))
im = ax.imshow(heat.values.astype(float), aspect='auto', interpolation='nearest')
ax.set_title('Extension 1: OOS R2 by Sub-Period')
ax.set_xticks(range(len(heat.columns)))
ax.set_xticklabels(heat.columns, rotation=45, ha='right')
ax.set_yticks(range(len(heat.index)))
ax.set_yticklabels(heat.index)
plt.colorbar(im, ax=ax, label='OOS R2')
plt.tight_layout()
plt.show()
```

```python
pars_sorted = pars.sort_values('delta_r2')
fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(pars_sorted.index, pars_sorted['delta_r2'], color='#9467bd')
ax.set_title('Extension 3: Parsimony Cost (delta R2 = phase5c - phase4)')
ax.set_ylabel('delta R2')
ax.tick_params(axis='x', rotation=45)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print('Degenerate month summary')
print(deg.to_string(index=False))
```
